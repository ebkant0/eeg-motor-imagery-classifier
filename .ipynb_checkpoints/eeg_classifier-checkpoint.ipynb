## Step 1: Load the Data

The PhysioNet EEG Motor Imagery dataset contains EEG recordings from 
109 subjects. Each subject performed and imagined opening and closing 
their left and right fists.

We load runs 6 and 10 which correspond to imagined left vs right hand 
movement — this is the core task of a brain computer interface.

```python
from mne.datasets import eegbci
from mne.io import concatenate_raws, read_raw_edf

# sets the data path explicitly so MNE never prompts for input
# points to where MNE already downloaded the data
mne.set_config('MNE_DATA', '/Users/chiebukaonwuka/mne_data')

# define which runs to load
# run 6 = imagined left vs right fist (first session)
# run 10 = imagined left vs right fist (second session)
runs = [6, 10]

# downloads data for subject 1 if not already downloaded
raw_fnames = eegbci.load_data(subjects=1, runs=runs, verbose=False)

# loads each file and concatenate into one continuous recording
# preload=True loads data into memory for faster processing
raw = concatenate_raws([read_raw_edf(f, preload=True, verbose=False) 
                        for f in raw_fnames])

# prints a summary of the data structure
# shows channels, sampling frequency, duration, and more
print(raw.info)
```

## Step 2: Explore the Raw Data

Before processing anything we need to understand what the raw EEG 
signal looks like. We'll check the duration, channel names, and 
plot the raw signal to see the brain activity directly.

```python
# prints total duration of the recording in seconds
duration = raw.times[-1]
print(f"Recording duration: {duration:.1f} seconds")

# prints total number of channels
print(f"Number of channels: {raw.info['nchan']}")

# prints sampling frequency
print(f"Sampling frequency: {raw.info['sfreq']} Hz")

# prints total number of time points
print(f"Total time points: {len(raw.times)}")

# prints first 10 channel names
print(f"\nFirst 10 channels: {raw.ch_names[:10]}")
```

## Step 3: Visualize the Raw EEG Signal

Now we plot the raw brain signal to see what EEG actually looks like 
before any processing. We expect to see oscillating waves across 
channels with occasional artifacts from eye blinks or movement.

```python
# cleans up channel names and removes trailing dots MNE adds
eegbci.standardize(raw)

# plot 5 seconds of raw EEG signal across all channels
# opens interactive window
raw.plot(
    duration=5,        # shows 5 seconds at a time
    n_channels=10,     # shows 10 channels at once
    scalings='auto',   # auto scale the amplitude
    title='Raw EEG Signal — Subject 1 Motor Imagery'
)
```
